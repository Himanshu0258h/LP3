# Uber Fare Prediction using Regression

### Problem Statement
Predict the price of an Uber ride from a given pickup point to the agreed drop-off location.

### Tasks
1. Pre-process the dataset.
2. Identify outliers.
3. Check correlation.
4. Implement Linear Regression and Random Forest Regression.
5. Evaluate and compare the models using R², RMSE and MAE.

**Dataset:** Uber Fares Dataset by Yasser H.  
Kaggle: https://www.kaggle.com/datasets/yasserh/uber-fares-dataset


## 1. Import Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error

sns.set_style("whitegrid")
print("Libraries imported successfully!")
```

## 2. Upload and Load Dataset

Download `uber.csv` from the Kaggle dataset and upload it using the cell below.

```python
uploaded = files.upload()

file_name = list(uploaded.keys())[0]
df = pd.read_csv(file_name)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
df.head()
```

## 3. Understand the Dataset

```python
print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nStatistical Summary:")
display(df.describe())
```

## 4. Data Pre-processing

The dataset contains unnecessary index/key columns, missing values, invalid coordinates, invalid fares and unrealistic passenger counts. The pickup timestamp is converted into useful numerical features.

We also calculate the approximate trip distance using the Haversine formula.

```python
# Remove unnecessary columns
df = df.drop(columns=["Unnamed: 0", "key"], errors="ignore")

# Convert pickup_datetime
df["pickup_datetime"] = pd.to_datetime(
    df["pickup_datetime"], errors="coerce", utc=True
)

# Remove missing values
df = df.dropna()

# Keep valid fare values
df = df[df["fare_amount"] > 0]

# Keep valid passenger counts
df = df[df["passenger_count"].between(1, 6)]

# Keep valid geographic coordinates
df = df[
    df["pickup_longitude"].between(-180, 180) &
    df["dropoff_longitude"].between(-180, 180) &
    df["pickup_latitude"].between(-90, 90) &
    df["dropoff_latitude"].between(-90, 90)
]

# Remove zero coordinates
df = df[
    (df["pickup_longitude"] != 0) &
    (df["pickup_latitude"] != 0) &
    (df["dropoff_longitude"] != 0) &
    (df["dropoff_latitude"] != 0)
].copy()

# Extract date/time features
df["pickup_hour"] = df["pickup_datetime"].dt.hour
df["pickup_day"] = df["pickup_datetime"].dt.day
df["pickup_month"] = df["pickup_datetime"].dt.month
df["pickup_weekday"] = df["pickup_datetime"].dt.dayofweek

print("Shape after preprocessing:", df.shape)
df.head()
```

## 5. Calculate Trip Distance

The Haversine formula calculates the approximate straight-line distance between pickup and drop-off coordinates.

```python
# Haversine distance
lat1 = np.radians(df["pickup_latitude"])
lat2 = np.radians(df["dropoff_latitude"])
lon1 = np.radians(df["pickup_longitude"])
lon2 = np.radians(df["dropoff_longitude"])

dlat = lat2 - lat1
dlon = lon2 - lon1

a = (
    np.sin(dlat / 2) ** 2
    + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2) ** 2
)

df["distance_km"] = 6371 * 2 * np.arcsin(np.sqrt(a))

# Remove zero-distance trips
df = df[df["distance_km"] > 0].copy()

print(df[["fare_amount", "distance_km"]].head())
```

## 6. Identify Outliers using IQR

The Interquartile Range (IQR) method identifies observations outside:

**Lower Bound = Q1 − 1.5 × IQR**  
**Upper Bound = Q3 + 1.5 × IQR**

```python
def iqr_bounds(series):
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    return q1 - 1.5 * iqr, q3 + 1.5 * iqr

fare_low, fare_high = iqr_bounds(df["fare_amount"])
distance_low, distance_high = iqr_bounds(df["distance_km"])

fare_outliers = ~df["fare_amount"].between(fare_low, fare_high)
distance_outliers = ~df["distance_km"].between(distance_low, distance_high)

print("Fare outliers:", fare_outliers.sum())
print("Distance outliers:", distance_outliers.sum())

# Visualize outliers
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.boxplot(y=df["fare_amount"], ax=axes[0])
axes[0].set_title("Fare Amount - Before Outlier Removal")

sns.boxplot(y=df["distance_km"], ax=axes[1])
axes[1].set_title("Distance - Before Outlier Removal")

plt.tight_layout()
plt.show()
```

```python
# Remove fare and distance outliers
df = df[
    df["fare_amount"].between(fare_low, fare_high) &
    df["distance_km"].between(distance_low, distance_high)
].copy()

print("Rows after removing outliers:", len(df))

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.boxplot(y=df["fare_amount"], ax=axes[0])
axes[0].set_title("Fare Amount - After Outlier Removal")

sns.boxplot(y=df["distance_km"], ax=axes[1])
axes[1].set_title("Distance - After Outlier Removal")

plt.tight_layout()
plt.show()
```

## 7. Check Correlation

Correlation shows the strength and direction of the linear relationship between numerical variables.

```python
features = [
    "pickup_longitude",
    "pickup_latitude",
    "dropoff_longitude",
    "dropoff_latitude",
    "passenger_count",
    "pickup_hour",
    "pickup_day",
    "pickup_month",
    "pickup_weekday",
    "distance_km"
]

correlation = df[features + ["fare_amount"]].corr()

plt.figure(figsize=(11, 8))
sns.heatmap(correlation, annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()

print("Correlation with fare_amount:")
display(
    correlation["fare_amount"]
    .sort_values(ascending=False)
    .to_frame("Correlation")
)
```

## 8. Prepare Data for Machine Learning

Target variable: `fare_amount`

Input features include geographic coordinates, passenger count, time features and calculated trip distance.

```python
X = df[features]
y = df["fare_amount"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

## 9. Linear Regression Model

```python
linear_model = LinearRegression()
linear_model.fit(X_train, y_train)

linear_pred = linear_model.predict(X_test)

linear_r2 = r2_score(y_test, linear_pred)
linear_rmse = np.sqrt(mean_squared_error(y_test, linear_pred))
linear_mae = mean_absolute_error(y_test, linear_pred)

print("Linear Regression Results")
print("R²   :", round(linear_r2, 4))
print("RMSE :", round(linear_rmse, 4))
print("MAE  :", round(linear_mae, 4))
```

## 10. Random Forest Regression Model

```python
rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)

rf_pred = rf_model.predict(X_test)

rf_r2 = r2_score(y_test, rf_pred)
rf_rmse = np.sqrt(mean_squared_error(y_test, rf_pred))
rf_mae = mean_absolute_error(y_test, rf_pred)

print("Random Forest Regression Results")
print("R²   :", round(rf_r2, 4))
print("RMSE :", round(rf_rmse, 4))
print("MAE  :", round(rf_mae, 4))
```

## 11. Compare the Models

```python
results = pd.DataFrame({
    "Model": ["Linear Regression", "Random Forest Regression"],
    "R2 Score": [linear_r2, rf_r2],
    "RMSE": [linear_rmse, rf_rmse],
    "MAE": [linear_mae, rf_mae]
})

display(results.round(4))

# Visual comparison
results.set_index("Model")[["R2 Score", "RMSE", "MAE"]].plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Model Performance Comparison")
plt.ylabel("Score")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 12. Actual vs Predicted Fare

```python
plt.figure(figsize=(7, 5))
plt.scatter(y_test, rf_pred, alpha=0.3)

min_val = min(y_test.min(), rf_pred.min())
max_val = max(y_test.max(), rf_pred.max())

plt.plot([min_val, max_val], [min_val, max_val], linestyle="--")
plt.xlabel("Actual Fare")
plt.ylabel("Predicted Fare")
plt.title("Random Forest: Actual vs Predicted Fare")
plt.show()
```

## Conclusion

The Uber fare dataset was pre-processed by removing invalid records, handling missing values, extracting date/time features and calculating trip distance.

Outliers were identified using the IQR method and removed. Correlation analysis was performed to understand relationships between the features and fare amount.

Two regression models were implemented:
- Linear Regression
- Random Forest Regression

The models were evaluated using **R², RMSE and MAE**. In general, the model with the **higher R² and lower RMSE/MAE** provides better prediction performance on the test data.
